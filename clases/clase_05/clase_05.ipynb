## ¿Qué son NumPy y pandas?

![](notas_files/320px-NumPy_logo_2020.svg.png)

Numpy es una biblioteca de Python de código abierto que se utiliza para la informática científica y proporciona una serie de características que permiten a un programador de Python trabajar con matrices y matrices de alto rendimiento.

![](notas_files/320px-Pandas_logo.svg.png)

Pandas es un paquete para la manipulación de datos que usa los objetos DataFrame de R (así como diferentes paquetes de R) en un entorno Python.

Tanto NumPy como pandas se usan a menudo juntos, ya que la biblioteca de pandas depende en gran medida de la matriz NumPy para la implementación de objetos de datos de pandas y comparte muchas de sus características. Además, pandas se basa en la funcionalidad proporcionada por NumPy. Ambas bibliotecas pertenecen a lo que se conoce como la pila SciPy, un conjunto de bibliotecas de Python utilizadas para la informática científica. 

#### Librerías utilizadas

- `pandas` (`pd`): librería para manipulación y análisis de datos en estructuras como DataFrames y Series.  
- `numpy` (`np`): librería para operaciones numéricas y manejo de arreglos multidimensionales, muy útil para cálculos matemáticos y estadísticos.


````python
import pandas as pd
import numpy as np 
````

### Matrices NumPy

NumPy le permite trabajar con matrices y matrices de alto rendimiento. Su objeto de datos principal es el ndarray, un tipo de matriz de N dimensiones que describe una colección de "elementos" del mismo tipo. Por ejemplo:

#### Creación de un arreglo con NumPy

- `np.array([1, 2, 3, 4, 5])`: crea un **ndarray** (arreglo de NumPy) con los elementos 1, 2, 3, 4 y 5.  
  - Los **ndarray** son estructuras de datos eficientes para operaciones numéricas vectorizadas.
  - Este comando define un arreglo unidimensional con los valores dados.


````python
np.array ([1, 2, 3, 4, 5]) # definiendo el ndarray
````

Los ndarrays se almacenan de manera más eficiente que las listas de Python y permiten vectorizar las operaciones matemáticas, lo que da como resultado un rendimiento significativamente mayor que con las construcciones de bucle en Python.

Las matrices NumPy permiten seleccionar elementos de matriz, operaciones lógicas, cortar, remodelar, combinar (también conocido como "apilar"), dividir, así como varios métodos numéricos (mínimo, máximo, media, desviación estándar, varianza y más). Todos estos conceptos se pueden aplicar a los objetos pandas, que amplían estas capacidades para proporcionar un medio mucho más rico y expresivo de representar y manipular datos que los que se ofrecen con las matrices NumPy.

## Series

La Serie es el componente principal de los pandas. Una serie representa una matriz indexada etiquetada unidimensional basada en el ndarray NumPy. 

> ``` s = pd.Series(data, index=index) ```

Como una matriz, una serie puede contener cero o más valores de cualquier tipo de datos. Se puede crear e inicializar una serie pasando un valor escalar, un ndarray NumPy, una lista de Python o un Dict de Python como parámetro de datos del constructor de series.

##### A partir de una serie

#### Creación de una Serie de pandas

- `a = pd.Series([1, 3, 5, np.nan, 6, 8])`: crea una **Serie** de pandas con los valores dados.  
  - Una Serie es un arreglo unidimensional que puede contener cualquier tipo de dato.  
  - `np.nan` representa un valor **faltante o nulo**.  
- `a`: muestra la Serie creada.


````python
a = pd.Series([1, 3, 5, np.nan, 6, 8])
a
````

###### Indice generado automáticamente

#### Acceso al índice de la Serie

- `a.index`: devuelve el **índice** de la Serie `a`.  
  - El índice es la "etiqueta" de cada elemento de la Serie.  
  - Por defecto, pandas asigna un rango de enteros comenzando en 0.


````python
a.index
````

##### A partir de un diccionario

#### Creación de una Serie a partir de un diccionario

- `d = {'b': 1, 'a': 0, 'c': 2}`: define un diccionario de Python con claves y valores.  
- `pd.Series(d)`: convierte el diccionario en una **Serie de pandas**.  
  - Las **claves** del diccionario se convierten en el **índice** de la Serie.  
  - Los **valores** del diccionario se convierten en los datos de la Serie.


````python
d = {'b': 1, 'a': 0, 'c': 2}
pd.Series(d)
````

##### A partir de una valor

#### Creación de una Serie con un valor constante

- `pd.Series(5., index=['a', 'b', 'c', 'd', 'e'])`: crea una **Serie de pandas** con **valor constante** 5.0 en todas las posiciones.  
  - `index=['a', 'b', 'c', 'd', 'e']` define las **etiquetas** del índice de la Serie.  
  - Cada elemento de la Serie tiene el mismo valor, útil para inicializaciones o asignaciones.


````python
pd.Series(5., index=['a', 'b', 'c', 'd', 'e'])
````

### Diferencias entre ndarrays y objetos de serie

Hay algunas diferencias que vale la pena señalar entre los objetos ndarrays y Series. En primer lugar, se accede a los elementos de las matrices NumPy por su posición entera, comenzando con cero para el primer elemento. Un objeto de la serie pandas es más flexible, ya que puede usar definir su propio índice etiquetado para indexar y acceder a elementos de una matriz. También puede utilizar letras en lugar de números o numerar una matriz en orden descendente en lugar de ascendente. En segundo lugar, alinear datos de diferentes Series y hacer coincidir etiquetas con objetos de Series es más eficiente que usar ndarrays, por ejemplo, tratar con valores perdidos. Si no hay etiquetas coincidentes durante la alineación, pandas devuelve NaN (no ningún número) para que la operación no falle.

## DataFrame

Un DataFrame es una estructura de datos etiquetada bidimensional con columnas de tipos potencialmente diferentes. Puede pensar en ello como una hoja de cálculo o una tabla SQL, o un dict de objetos Series. Generalmente es el objeto de pandas más utilizado. Al igual que Series, DataFrame acepta muchos tipos diferentes de entrada:

- 1D Dict de ndarrays, listas, dict o series

- 2-D numpy.ndarray

- Ndarray estructurado

- Series

- Otro DataFrame


### Creación de un Data-Frame (df)


#### Vía Numpy


#### Creación de un rango de fechas

- `dates = pd.date_range('20130101', periods=6)`: genera un **rango de fechas** comenzando el 1 de enero de 2013.  
  - `periods=6` indica que se generarán **6 fechas consecutivas**.  
  - El resultado es un objeto `DatetimeIndex`, útil para trabajar con series temporales en pandas.  
- `dates`: muestra el rango de fechas creado.


````python
dates = pd.date_range('20130101', periods=6)
dates
````

#### Creación de un DataFrame con valores aleatorios

- `df = pd.DataFrame(np.random.randn(6, 4), index=dates, columns=['A', 'B', 'C', 'D'])`: crea un **DataFrame** de pandas.  
  - `np.random.randn(6, 4)` genera un arreglo de NumPy de **6 filas y 4 columnas** con valores aleatorios siguiendo una distribución normal.  
  - `index=dates` asigna el **índice** del DataFrame usando el rango de fechas creado anteriormente.  
  - `columns=['A', 'B', 'C', 'D']` asigna los **nombres de las columnas**.  
- `df`: muestra el DataFrame resultante.


````python
df = pd.DataFrame(np.random.randn(6, 4),
                  index=dates,
                  columns=['A', 'B', 'C', 'D'])
df
````

#### Vía diccionarios de Python


#### Creación de un DataFrame con diferentes tipos de datos

- `df2 = pd.DataFrame({...})`: crea un **DataFrame** especificando cada columna con distintos tipos de datos.  
  - `'A': 1.` → columna de tipo **float** con valor constante 1.0.  
  - `'B': pd.Timestamp('20130102')` → columna de tipo **fecha/hora** con la misma fecha para todas las filas.  
  - `'C': pd.Series(1, index=list(range(4)), dtype='float32')` → Serie de 4 elementos, tipo **float32**, valor constante 1.  
  - `'D': np.array([3] * 4, dtype='int32')` → arreglo NumPy de enteros (int32) con valor 3 repetido 4 veces.  
  - `'E': pd.Categorical(["test", "train", "test", "train"])` → columna categórica con 2 categorías: "test" y "train".  
  - `'F': 'foo'` → columna de tipo string con valor constante "foo".  
  - `'G': [1,2,1,2]` → columna de enteros usando una lista de Python.  
- `df2`: muestra el DataFrame resultante, demostrando que pandas puede manejar múltiples tipos de datos en un mismo DataFrame.


````python
df2 = pd.DataFrame({
    'A': 1.,
    'B': pd.Timestamp('20130102'),
    'C': pd.Series(1, index=list(range(4)), dtype='float32'),
    'D': np.array([3] * 4, dtype='int32'),
    'E': pd.Categorical(["test", "train", "test", "train"]),
    'F': 'foo',
    'G': [1,2,1,2]
})
df2
````

### Los datos faltantes

* None: el dato faltante Pythonico

* NaN (Not a number): representación de un número faltante reconocido por todo los sistemas que usan el estandar de de IEEE de coma flotante

> Ver https://jakevdp.github.io/PythonDataScienceHandbook/03.04-missing-values.html


#### Serie con valores faltantes

- `pd.Series([1, np.nan, 2, None])`: crea una **Serie de pandas** que incluye **valores faltantes**.  
  - `np.nan` y `None` son tratados por pandas como **NaN** (Not a Number).  
  - Esto es útil para representar datos incompletos o ausentes en análisis y cálculos.


````python
pd.Series([1, np.nan, 2, None])
````

#### Operación con valor faltante

- `np.nan + 1`: realiza una operación aritmética con **NaN**.  
  - El resultado es `NaN`, porque cualquier operación con un valor faltante sigue siendo **desconocida**.  
  - Esto es útil para entender cómo pandas y NumPy manejan datos faltantes en cálculos.


````python
np.nan + 1
````

#### Comparación entre NaN y otro valor

- `np.nan == '2'`: compara un **valor faltante (NaN)** con un valor de tipo string `'2'`.  
  - El resultado es `False`.  
  - En general, **NaN no es igual a nada**, ni siquiera a otro NaN (`np.nan == np.nan` también devuelve `False`).  
  - Esto refleja que NaN representa un valor desconocido o indefinido.


````python
np.nan == '2'
````

#### Comparación de NaN con un número

- `np.nan > 0`: compara un **valor faltante (NaN)** con un número.  
  - El resultado es `False`.  
  - Cualquier comparación lógica con NaN (**>, <, ==, !=**) devuelve `False` excepto `!=` que devuelve `True`.  
  - Esto ocurre porque NaN representa un valor desconocido, por lo que no se puede determinar si es mayor o menor que un número.


````python
np.nan > 0
````

## Importar datos


#### Lectura de un archivo CSV con pandas

- `pd.read_csv(...)`: lee un archivo CSV y lo convierte en un **DataFrame de pandas**.  
- Parámetros principales usados en este ejemplo:  
  - `filepath_or_buffer='../../dataset/censo2010/hogar.csv'`: ruta al archivo CSV.  
  - `delimiter=','`: caracter que separa los campos (puede ser `','`, `';'`, `'|'`, `'\t'`).  
  - `header=0`: indica que la **primera fila** contiene los nombres de las columnas.  
  - `names=None`: permite definir **nombres personalizados** para las columnas (cuidado si se usa junto con `header`).  
  - `index_col=0`: define qué columna usar como **índice** del DataFrame.  
  - `usecols=None`: permite seleccionar solo algunas columnas, ej. `[0,1,2]` o `['foo','bar']`.  
  - `dtype=None`: define tipos de datos específicos por columna, ej. `{'a': np.int32, 'b': str}`.  
  - `skiprows=None`: saltear filas al inicio del archivo.  
  - `skipfooter=0`: saltear filas al final del archivo.  
  - `nrows=None`: leer solo un número limitado de filas.  
  - `decimal='.'`: caracter usado como separador decimal (ej. ',' en algunos países europeos).  
  - `quotechar='"'`: caracter usado para reconocer texto entre comillas.  
  - `encoding=None`: útil para manejar acentos, ñ u otros caracteres especiales.  

- El resultado es un **DataFrame** listo para análisis y manipulación de datos.


```python
pd.read_csv(
    '../../dataset/censo2010/hogar.csv',  # file path
    delimiter=',',  # delimitador ',',';','|','\t'
    header=0,  # número de fila como nom de col
    names=None,  # nombre de las columnas (ojo con header)
    index_col=0,  # que col es el índice
    usecols=None,  # que col usar. Ej: [0, 1, 2], ['foo', 'bar', 'baz']
    dtype=None,  # Tipo de col {'a': np.int32, 'b': str} 
    skiprows=None,  # saltear fil al init
    skipfooter=0,  # saltear fil al final
    nrows=None,  # n de fil a leer
    decimal='.',  # separador de decimal. Ej: ',' para EU dat
    quotechar='"',  # char para reconocer str
    #encoding=None,  # para los acentos, ñ, etc
)
```

- Se pueden seleccionar una o varias columnas usando **corchetes** `[]` o `.loc[]`.  

- Ejemplos prácticos:
  ```python

> Operator Overloading
Ciertos operadores poseen un comportamiento diferenciado de a cuerdo a que objetos se los aplica. Por ejemplo, al indexar con $df[*]$, su comportamiento dependerá de si es:
* string: retornará una columna como una serie return a column as a Series
* lista de strings: retornará columnas en forma de DataFrame

#### Condiciones Múltiples


##### Usando &, | , ~
##### Filtrado de filas usando operadores lógicos `&`, `|` y `~`

- En pandas, se pueden combinar condiciones usando **operadores bit a bit**:  
  - `&` → **AND** lógico: devuelve `True` si **ambas condiciones** son verdaderas.  
  - `|` → **OR** lógico: devuelve `True` si **al menos una condición** es verdadera.  
  - `~` → **NOT** lógico: invierte la condición (`True` → `False` y viceversa).  

- Ejemplos prácticos:
  ```python
  df_in[(df_in['ALGUNBI'] > 0) & (df_in['PROP'] > 1000)]  # filas donde ALGUNBI>0 y PROP>1000
  df_in[(df_in['ALGUNBI'] > 0) | (df_in['PROP'] > 1000)]  # filas donde ALGUNBI>0 o PROP>1000
  df_in[~(df_in['ALGUNBI'] > 0)]  # filas donde ALGUNBI no es mayor a 0

## Unstack
#### Transformar filas en columnas (`unstack`)

- `unstack()` es la operación inversa de `stack()`:  
  - **Convierte un nivel del índice (filas) en columnas**.  
  - Útil para pasar datos de un **formato largo** a un **formato ancho**.

- Ejemplo práctico:
  ```python

      'A': [1, 2],

      'C': [5, 6]

#### Ejemplo de `unstack()` aplicado a un DataFrame con Multi-Índice

- Código:
  ```python

      .groupby(['PROP','ALGUNBI'])
      .mean()
  )
  df.unstack()

- Paso a paso:

Primero se agrupa df_in por ['PROP','ALGUNBI'] y se calcula el promedio de 'TOTPERS'.

El resultado (df) es un DataFrame con Multi-Índice en las filas (PROP, ALGUNBI).

Con .unstack(), el segundo nivel del índice (ALGUNBI) se convierte en columnas.

- Resultado:

Ahora cada valor distinto de ALGUNBI aparece como columna.

'PROP' se mantiene como índice en las filas.

## Inner join

Esta cláusula busca coincidencias entre 2 tablas, en función a una columna que tienen en común. De tal modo que sólo la intersección se mostrará en los resultados.

![](notas_files/INNER_JOIN.webp)

Veamos un Ejemplo

![](notas_files/join-inner_2.png)

````python
#inner join in python pandas

inner_join_df= pd.merge(x, y, on='key', how='inner')
inner_join_df 
````

## Left Join

A diferencia de un INNER JOIN, donde se busca una intersección respetada por ambas tablas, con LEFT JOIN damos prioridad a la tabla de la izquierda, y buscamos en la tabla derecha. Si no existe ninguna coincidencia para alguna de las filas de la tabla de la izquierda, de igual forma todos los resultados de la primera tabla se muestran.

![](notas_files/LEFT_JOIN.webp)

Veamos un Ejemplo

![](notas_files/left_joint.png)

````python
left_join_df = pd.merge(x, y, on='key', how='left')
left_join_df
````

## Right Join

En el caso de RIGHT JOIN la situación es muy similar, pero aquí se da prioridad a la tabla de la derecha.

![](notas_files/RIGHT_JOIN.webp)

Veamos un Ejemplo

![](notas_files/right_joint.png)

````python
right_join_df = pd.merge(x, y, on='key', how='right')
right_join_df
````

## Full Join

Mientras que LEFT JOIN muestra todas las filas de la tabla izquierda, y RIGHT JOIN muestra todas las correspondientes a la tabla derecha, FULL OUTER JOIN (o simplemente FULL JOIN) se encarga de mostrar todas las filas de ambas tablas, sin importar que no existan coincidencias (usará NULL como un valor por defecto para dichos casos).

![](notas_files/FULL_JOIN.webp)

Veamos un Ejemplo

![](notas_files/outer_joint.png)

````python
outer_join_df = pd.merge(x, y, on='key', how='outer')
outer_join_df
````
